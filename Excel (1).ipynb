1.	What is the purpose of a PivotTable in Excel?

Answer-Purpose of a PivotTable
•	Summarize large datasets into meaningful reports
•	Calculate totals, averages, counts, and percentages
•	Compare data across categories (e.g., Region, Product, Month)
•	Rearrange data dynamically without changing the original dataset
•	Create interactive reports for analysis

4.Explain the difference between Sorting and Filtering with an example from this dataset.

Ans -Sorting is used to arrange data in a specific order, either ascending or descending, based on the values in one or more columns. It changes the order of the records in the dataset.
Example from Dataset:
If the Sales column is sorted from highest to lowest, Excel will rearrange all rows so that the fruit sale with the highest sales amount appears first and the lowest appears last.
Purpose of Sorting:
•	To rank data from highest to lowest or vice versa
•	To organize records alphabetically, numerically, or by date
•	To make comparison easier
Filtering is used to display only the records that meet specific criteria while temporarily hiding the remaining data. It does not change the order of the dataset.
Example from Dataset:
If the Region column is filtered to show only East, Excel will display only the fruit sales records from the East region and hide all other regions.
Purpose of Filtering:
•	To focus on specific data
•	To analyze selected records only
•	To hide unnecessary information temporarily

8.	Why are Dashboards important for decision-making in business analysis?

Ans-Dashboards are important in business analysis because they present key data and performance metrics in a visual, organized, and interactive format. They help managers and analysts understand complex data quickly and make informed decisions.
Importance of Dashboards
1.	Provide Quick Insights
Dashboards summarize large amounts of data into charts, tables, and KPIs, allowing users to understand performance at a glance.
2.	Support Better Decision-Making
Visual reports help management identify trends, problems, and opportunities faster.
3.	Improve Data Visualization
Charts and graphs make data easier to interpret compared to raw numbers.
4.	Enable Real-Time Monitoring
Dashboards can be updated dynamically to reflect the latest business performance.
5.	Allow Interactive Analysis
Tools like slicers and filters let users explore data from different perspectives.
6.	Save Time
Instead of reviewing multiple reports, users can access all important information in one place.
